```python
import pandas as pd
import numpy as np
```

```python

df = pd.read_csv('student-por.csv')
print('Dataset shape (rows, columns):', df.shape)
print("Column name:", df.columns.tolist())
print('First 5 rows of the dataset:')
print(df.head())
print(df.describe())

print("Data types in and null info")
print(df.info()) # qanaqa malumot ekaligini blishmz uchun
```

```python

print("Average scores by subject")
df[["G1", "G2", "G3"]].mean()
```

```python
print("Average scores by subject")
df[["G1", "G2", "G3"]].max() # baland baxo
```

```python
print("Average scores by subject")
df[["G1", "G2", "G3"]].min() # pas baxo
```

```python
#Correlation
print("Correlation")
df[["G1", "G2", "G3"]].corr()
```

```python
df["total score"] = df[["G1", "G2", "G3"]].sum(axis=1)
df["average score"] = df["total score"] / 3
df["passed"] = np.where(df["average score"] >= 10, "Yes", "No")

print("New columns preview (total, average, passed):")
print(df[["total score", "average score", "passed"]].head())
```

```python
test_prep_group = df.groupby("schoolsup")[["G1", "G2", "G3"]].mean()
print("student score by school support:")
print(test_prep_group)
```

```python
# Gender-based average
gender_avg = df.groupby("sex")[["G1", "G2", "G3"]].mean()
print("Average score by gender:")
print(gender_avg)
```

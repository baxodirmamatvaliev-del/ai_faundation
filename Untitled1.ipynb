```python
import numpy as np

arr = np.array([1,3,5,7])
print("numpy array", arr)
```

```python
arr2 = np.array([[1,3,5,7], [4,6,8,9]])
print("numpy array2", arr2)
```

```python
# dtype()
a = np.array([1,2,3])
print(a.dtype)
```

```python
# reshape()
a = np.array([1,2,3,4,5,6])
b = a.reshape(2,3)
print("shaped_array",b)
```

```python
#  astype() — ma’lumot turini o‘zgartirish
a = np.array([1, 2, 3])
b = a.astype(float)
print("float:", b)
```

```python
#  zeros() — nollardan iborat array, # 2. ones() — birlardan iborat array yaratadi
z = np.zeros((2, 3))
o = np.ones((2, 3))
print("zeros:", z)
print("ones:", o)
```

```python
# ndim — massiv o‘lchamlari soni
a = np.array([[1, 2], [3, 4]])
print("ndim:", a.ndim)
```

```python
# arange() — oraliqdagi sonlar
a = np.arange(1, 6)
print("arange:", a)
```

```python
#  sum(), mean(), max(), min()
a = np.array([1, 2, 3, 4])
print("sum:", a.sum()) # teskari sanoqdan boshlab qiymatni chiqarib beradi
print("mean:", a.mean()) # sonlardan kelib chiqib ortachasini anuqlab beradi 2.5
print("max:", a.max()) # sonlar ichidan end yuqorisini topib beradi "4"
print("min:", a.min())# sonlar ichidan end kichigini topib beradi "1"
```

```python
# flatten() —  bizga 2D... 1D ga keltirib vector qlib beradi
import numpy as np
a = np.array([[1, 2], [3, 4]])
print("flatten:", a.flatten())
```

```python
#  T — (transpozitsiya)
a = np.array([[1, 2], [3, 4]])
print("T:", a.T)
```

```python
# dot() — matritsalarni ko‘paytirish
a = np.array([[1, 2], [3, 4]])
b = np.array([[2, 0], [1, 2]])
print("dot/multiplication:", a.dot(b))
```

```python

# random.rand() — 0 va 1 oralig‘idagi tasodifiy sonlar
a = np.random.rand(2, 3)
print("random rand:", a)
```
